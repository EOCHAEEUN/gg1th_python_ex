```python
def comMul(num1, num2):
    result=[]
    #pass
    #1~100까지 공배수 공하기 
    #num1*num2  ->  최소 공배수
    for i in range(1,101):
        if i % num1 ==0 and i % num2 ==0:
            result.append(i)
    return result

print(comMul(num1, num2))
    
```
