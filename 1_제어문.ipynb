```python

nums=[273,103,5,32,65,9,72,800,99,121]
len(nums)
```

```python
#100이상의 숫자만 출력 
for num in nums:
    if num >= 100:
        print(f"100d이상의 숫자 :{num}")
```

```python
#100이상의 숫자 100 미만인 숫자를 따로 새로운 리스트에 넣어서 가각 리스트로 출력
lower100=[]
higher100=[]
for num in nums:
    if num >=100:
        higher100.append(num)
    else:
        #100미만 리스트 추가 
        lower100.append(num)

print(f"-100이상의 수:{higher100}")
print(f"-100미만의 수:{lower100}")
```

## >단축키
-text 셀-> code 셀: [esc -> Y] 
-code 셀-> text 셀: [esc -> x]
-실행: ctrl +ent
-실행 + 다음셀로 커서 이동 :shit + ent
-셀 삭제 : esc -> dd  **가장많이 쓰는 단축키 **
-셀 삭제 후 복원 (되돌리기): z 
-현재 셀의 위 쪽에 셀추가 : esc ->a 
-현재 셀의 아래 쪽에 셀추가 : esc ->b

```python
#nums의 숫자가 각각 몇번 반복하는지 딕트 로만들어서 출력

nums =[1,2,6,8,4,3,2,1,9,5,4,9,7,2,1,3,5,4,8,9,7,2,3]
counter = {}
#counter[1]=3 -> {1:3}

for num in nums:
    #키값 판단 
    if num in counter:
        counter[num] += 1 #이미 있는 숫자면  +1
    else:
        counter[num] = 1 # 처음 나왔으면 키값 , 초기 1 

print(counter)        
```

```python
#1~100까지 중 소수 구하기 

decimal = []

for num in range(2, 101):
    if num ==2:
        decimal.append(num)
    deno = num - 1

    for i in range(2, deno+1):
        if num % i == 0:
            break

        elif i == deno:
            decimal.append(num)

print(decimal)
```
